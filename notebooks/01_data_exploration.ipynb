# 01 - Data Exploration
Run from the `notebooks/` folder after placing CSVs in `data/raw/`.

```
import sys; sys.path.insert(0, '../src')
import pandas as pd, seaborn as sns, matplotlib.pyplot as plt
from preprocessing import load_data, clean, encode_labels, class_balance
df = clean(load_data('../data/raw'))
df.head()
```

```
y = encode_labels(df)
print(class_balance(y))
df['Label'].value_counts().plot(kind='bar', title='Traffic categories'); plt.show()
```

```
num = df.select_dtypes('number')
plt.figure(figsize=(8,6)); sns.heatmap(num.corr(), cmap='coolwarm'); plt.title('Feature correlations'); plt.show()
```

```
sns.boxplot(data=df, x='Label', y='Total Fwd Packets'); plt.yscale('log'); plt.xticks(rotation=30); plt.show()
```
